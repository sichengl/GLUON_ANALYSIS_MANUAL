```python
import numpy as np
import h5py
import gvar as gv
import lsqfit
import matplotlib.pyplot as plt

f=h5py.File("bareM_TXTXpTYTYm2XYXY_symmetric_tgf40_pf-1_0_5_q2_0_0_tsep5-10_svd1e-07_dEf5.h5","r")

for name in f:
    print(name)
Ai_jk = f["dEi_jk"]
Af_jk = f["dEf_jk"]
print(Ai_jk.shape)

def jk(data,jk_axis=0):
    n = data.shape[jk_axis]
    print(n)
    data = np.moveaxis(data, jk_axis, 0)
    print(f"data has shape {data.shape}")
    mean = np.mean(data,axis=0)
    print(f"mean has shape {mean.shape}")
    
    err = np.sqrt((n-1.0)/n * np.sum((data-mean)**2,axis=0))
    return mean, err

Ai_jk_mean, Ai_jk_err = jk(Ai_jk)
Af_jk_mean, Af_jk_err = jk(Af_jk)


print(f"Ai_jk_mean, {Ai_jk_mean}")
print(f"Ai_jk_err, {Ai_jk_err}")

print(f"Af_jk_mean, {Af_jk_mean}")
print(f"Af_jk_err, {Af_jk_err}")

def make_prior():
     
    prior = gv.BufferDict()
    prior["center"] = gv.gvar(0,1)
    #prior["add"] = gv.gvar(0,1)
    return prior

def model(p):
    #print(f"add is {p["add"]}")
    return [p["center"]]*100

Q_list = []
Chi2_list = []

for run in range(0,1000):
    normal_vars = np.random.normal(loc=0.0, scale=1.0, size=100)
    
    fit = lsqfit.nonlinear_fit(data=gv.gvar(normal_vars,np.eye(100) ), fcn=model, svdcut=1e-12,maxit=2000)
    print(f"chi2/dof is {fit.chi2/len(normal_vars)}")
    Q_list.append(fit.Q)
    Chi2_list.append(fit.chi2)

plt.hist(Q_list, bins=np.linspace(0,1,100)) 
```
